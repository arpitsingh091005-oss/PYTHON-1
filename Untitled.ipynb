```python
# sort()
marks = [34,67,65,87,89,43]
marks.sort() # ascending order
print(marks)
marks.sort(reverse =True) # descending order
print(marks)
```

```python
# EXAMPLE 1 :
students = {
    "Rahul":75,
    "Amit":32,
    "Priya":34,
    "Neha":39,
    "Rohit":55
}
for name, marks in students.items():
    if(marks>40):
        if(marks>=80):
            print(name," Excellent")
        elif(marks>=60):
            print(name," good")
        elif(marks>=40):
            print(name," Average")
    else:
        print(name," Fail")
```

```python
# EXAMPLE 3
products = {
    "Laptop": 10,
    "Mouse": 3,
    "Keyboard": 8,
    "Monitor": 7,
    "Printer": 2
}

for product, stock in products.items():
    if stock == 0:
        print(product, "out of stock")
    elif stock <= 5:
        print(product, "low stock")
    else:
        print(product, "sufficient stock")
```

```python
# EXAMPLE 4
employees = {
    "Rahul": {"salary": 60000, "department": "IT"},
    "Amit": {"salary": 40000, "department": "HR"},
    "Priya": {"salary": 70000, "department": "IT"},
    "Neha": {"salary": 45000, "department": "Finance"}
}

for name, data in employees.items():
    salary = data["salary"]
    department = data["department"]

    if salary >= 50000:
        if department == "IT":
            print(name, "Senior IT Employee")
        elif department == "HR":
            print(name, "Senior HR Employee")
        else:
            print(name, "Senior Employee")
    else:
        print(name, "Junior Employee")
```

```python
# EXAMPLE 5
data = {
    "a": 10,
    "b": 15,
    "c": 25,
    "d": 30,
    "e": 40
}

for key, value in data.items():
    if value > 20 and value % 2 == 0:
        print(value, "Large even")
    elif value % 2 == 0:
        print(value, "Even")
    else:
        print(value, "Odd")
```

```python
# EXAMPLE 6
products = {
    "Mobile": 15000,
    "Mouse": 500,
    "Laptop": 60000,
    "Keyboard": 12000
}
for product,price in products.items():
    if(price >= 1000):
        if(price >= 5000):
            print(product, "Premium product")
        elif(price < 1000):
            print(product, "Affordable product")
    else:
        print(product, "Expensive product")
            
```

```python
# EXAMPLE 7
students = {
    "Rahul": {"attendance": 85, "Marks": 70},
    "Amit": {"attendance": 60, "Marks": 80},
    "Priya": {"attendance": 90, "Marks": 45},
    "Neha": {"attendance": 78, "Marks": 55}
}

for name, data in students.items():
    attendance = data["attendance"]
    marks = data["Marks"]

    if attendance >= 75:
        if marks >= 40:
            print(name, marks, "Eligible")
        else:
            print(name, marks, "Failed in exam")
    else:
        print(name, "Not eligible")
```

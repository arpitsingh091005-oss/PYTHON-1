```python
x = 10
y = 10

print(x == y)
print(x is y) #Identity operator
print(id(x))
print(id(y))
```

```python
x = [30,50]
y = x
print(id(x))
print(id(y))
x.append(33)
print(x)
print(y)
print(x is y)
print(x == y)
```

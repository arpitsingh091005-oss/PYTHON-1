```python
def show():
    x=20 # Local Variable
    print("Local Variable: ",x)
show()    
```

```python
x=30 # GLOBAL VARIABLE
def show():
    x=20 # LOCAL VARIABLE
    print("Local Variable: ",x) # 20
show()

print("Global Variable: ",x) #30
```

```python
x=30 # Global Variable
def show():
    global x
    print("Global Variable: ",x)
show()
print("Global Variable: ",x)

def display():
    global x
    print("Global Variable: ",x)
display()    
```

```python
Balance=3000 # GlobalVariable
def show():
    global Balance
    deposit=1000
    print("Deposit Balance: ",deposit)

    Balance=Balance+deposit

show()
print("Current Balance after deposit: ",Balance)
```

```python
balance=3000
def show():
    deposit=1000
    print("Deposit balance: ",deposit)
    print("Initial balance was: ", globals() ["balance"])

show()
print("Current balance after deposit: ",balance)
```

```python
# Exercise 2:
def calculateBill(price,discount=0):
    discount_amount=price*discount/100
    bill=price-discount_amount

    print("Original Price: ",price)
    print("Discount : ",discount, "%")
    print("Discount Amount: ",discount_amount)
    print("Final Bill Amount: ",bill)
    print("========================")


calculateBill(20000)
calculateBill(10000,10)
calculateBill(30000,20)
```

```python
def Employeedetails(**kwargs):
    print(kwargs['name'])
    print(kwargs['Role'])
    print(kwargs['Department'])

    
Employeedetails(name="Mohit",Role="Developer",Department="IT")
```

```python
def Employeedetails(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

    print("====================")
Employeedetails(name="Mohit",Role="Developer",Department="IT")
Employeedetails(name="Mohit",Department="IT")
Employeedetails(name="Mohit")
```

```python
def studentresult(name,*marks,**details):
    print("student name: ",name)
    print("student marks: ",marks)

    if len(marks)>0:
        total=sum(marks)
        avg=total/len(marks)
        print("Total marks: ",total)
        print("Avg marks: ",avg)

    print("==========Additional information========")
    for key,value in details.items():
        print(key, ":" , value)

studentresult("Amit",50,45,70,87,97, roll=101,sem="1ST")
    
```

```python
fruits = ["apple","banana","cherry","kiwi","mango"]
newlist = []
for x in fruits:
    if "a" in x:
        newlist.append(x)

print(newlist)        
```

```python
# list comprehension
fruits = ["apple","banana","cherry","kiwi","mango"]
newlist = [x for x in fruits if "a" in x]
print(newlist)
```

```python
numbers = [20,55,30,80,45,90,10]
newlist = [x for x in numbers if x>50]
print(newlist)
```

```python
numbers = [10,25,35,60,45,15,50,70]
newlist = [x for x in numbers if 20<x<50]
print(newlist)
```

```python
numbers = [10,15,20,30,45,50,60,75]
newlist = [x for x in numbers if x % 3==0 and x % 5 == 0]
print(newlist)
```

```python
names = ["Amit","Rahul","Neha","Rohit","Arjun","Sneha"]
newlist = [x for x in  names  if len(names)>4]
print(newlist)
```

```python
names = ["Amit","Rahul","Ankit","Arpit","Devansh","Sneha"]
newlist = [name for name in names if name.startswith('A')] 
print(newlist)
```

```python
text = "programming"
newlist = [char for char in text if char in "aeiou"]
print(newlist)
```

```python
Data = "Python123Programming456"
newlist = [digit for digit in Data if digit in "123"]
print(newlist)
```

```python

numbers = [10, 15, 20, 25, 30]

newlist = ["Odd" if x % 2 == 0 else "Even" for x in numbers]

print(newlist)
```

```python
numbers = [10,-5,20,-8,30,-2]
newlist = [x  if x>0 else 0 for x in numbers]
print(newlist)
```

```python
price = [500,1200,800,2000,1500]
newlist = [ x*.90 if x>1000 else x*.95 for x in price]
print(newlist)
```

```python
employees = [
    {"name":"Amit","Salary":45000},
    {"name":"Rahul","Salary":60000},
    {"name":"Neha","Salary":75000},
    {"name":"Rohit","Salary":35000}
]
newlist = [ex['name'] for ex in employees if ex['Salary']>50000]
print(newlist)
```

```python
# E-Commmerce:
prices = [499,1299,799,2499,999,1599]
newlist = [ex for ex in prices if ex>1000]
print(newlist)
```

```python
prices = [500,1200,2500,3000,1500,5000]
newlist = [ex*0.80 if ex>2000 else 0 for ex in prices]
print(newlist)
```

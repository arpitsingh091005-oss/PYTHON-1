```python
# Tuple

data = (10,20,30,40,50)
print(type(data))
```

```python
data3 = (10)
print(type(data3))
data3 = (10,)
print(type(data3))
```

```python
data = (10,20,30,40,50)
print(data[3])
print(data[-1])
print(data[0:4])
```

```python
data = (10,20,30,40,50)
print(len(data))
print(sum(data))
print(data.index(20))
print(data.count(20))
```

```python
data = (10,20,30,60,80,40,50)
# data.sort() output: AttributeError: 'tuple' object has no attribute 'sort'
#Meaning
newdata =tuple(sorted(data))
print(data)
print(newdata)
```

```python
data = ("Arpit","MCA",24,False,24)   # Tuple Packing
name, course,age,attendance,marks = data   # Tuple Unpacking
print(name)
```

```python
data = ("Arpit","MCA",24,False,24)   #Tuple Packing
name, *restRecord=data
print(name)
print(restRecord)
```

```python
data = ("Arpit","MCA",24,False,24)  # Tuple Packing

name, *restRecord,marks=data
print(name)
print(restRecord)
print(marks)
```
